```python
import pandas as pd
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Create a synthetic Students Performance dataset with the official columns
n_students = 1000
genders = np.random.choice(['female', 'male'], size=n_students)
groups = np.random.choice(['group A', 'group B', 'group C', 'group D', 'group E'], size=n_students)
education = np.random.choice([
    "associate's degree", "some college", "high school", "some high school", "bachelor's degree", "master's degree"
], size=n_students)
lunch = np.random.choice(['standard', 'free/reduced'], size=n_students)
prep = np.random.choice(['none', 'completed'], size=n_students)

math_score = np.clip(np.random.normal(66, 15, size=n_students).astype(int), 0, 100)
reading_score = np.clip(np.random.normal(69, 14, size=n_students).astype(int), 0, 100)
writing_score = np.clip(np.random.normal(68, 15, size=n_students).astype(int), 0, 100)

data = {
    'gender': genders,
    'race/ethnicity': groups,
    'parental level of education': education,
    'lunch': lunch,
    'test preparation course': prep,
    'math score': math_score,
    'reading score': reading_score,
    'writing score': writing_score
}

df = pd.DataFrame(data)

# Display first five rows
display(df.head())
```

```python
# Dataset information
df.info()

# Check missing values
print(df.isnull().sum())

# Statistical summary
df.describe()
```

```python
x = df["math score"]
y = df["reading score"]
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(x, y, color="blue", alpha=0.6)

plt.title("Relationship Between Math and Reading Scores")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.grid(True)
plt.show()
```

```python
correlation = df["math score"].corr(df["reading score"])

print("Correlation:", correlation)
```

```python
m, b = np.polyfit(x, y, 1)

plt.figure(figsize=(8, 5))

plt.scatter(x, y, color="blue", alpha=0.5, label="Students")
plt.plot(x, m*x + b, color="red", label="Trend Line")

plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")

plt.legend()
plt.grid(True)
plt.show()
```
